```python
import numpy as np
import pandas as pd
import joblib
```

```python
heart_model = joblib.load("../models/trained_models/heart_RandomForest.pkl")

diabetes_model = joblib.load("../models/trained_models/diabetes_RandomForest.pkl")

cancer_model = joblib.load("../models/trained_models/cancer_RandomForest.pkl")

cancer_df = pd.read_csv("../data/raw/breast_cancer.csv")
```

Create Unified Prediction Function

```python
def predict_disease(disease_type, input_data):

    input_data = np.array(input_data).reshape(1, -1)

    if disease_type == "heart":
        model = heart_model
        prediction = model.predict(input_data)[0]

    elif disease_type == "diabetes":
        model = diabetes_model
        prediction = model.predict(input_data)[0]

    elif disease_type == "cancer":
        model = cancer_model
        prediction = model.predict(input_data)[0]

    else:
        return "Invalid disease type"

    return prediction
```

Test Heart Prediction

```python
heart_sample = [
    63, 1, 3, 145, 233, 1, 0, 150, 0, 2.3, 0, 0, 1
]

result = predict_disease("heart", heart_sample)

print("Heart Disease Prediction:", result)
```

Test Diabetes Prediction

```python
diabetes_sample = [
    6, 148, 72, 35, 0, 33.6, 0.627, 50
]

result = predict_disease("diabetes", diabetes_sample)

print("Diabetes Prediction:", result)
```

Test Cancer Prediction

```python
cancer_df = pd.read_csv("../data/raw/breast_cancer.csv")

cancer_sample = cancer_df.iloc[0].drop("target").values

result = predict_disease("cancer", cancer_sample)

print("Cancer Prediction:", result)
```

Feature Mapping

```python
# Heart Mapping

gender_map = {
    "male": 1, "Male": 1, "MALE": 1, "m": 1, "M": 1,
    "female": 0, "Female": 0, "FEMALE": 0, "f": 0, "F": 0
}

cp_map = {
    "typical angina": 0, "Typical Angina": 0, "typical": 0,
    "atypical angina": 1, "Atypical Angina": 1, "atypical": 1,
    "non-anginal pain": 2, "Non-Anginal Pain": 2, "non-anginal": 2,
    "asymptomatic": 3, "Asymptomatic": 3, "no pain": 3, "none": 3
}

yes_no = {
    "yes": 1, "Yes": 1, "YES": 1, "y": 1, "Y": 1, "true": 1, "True": 1, "1": 1,
    "no": 0, "No": 0, "NO": 0, "n": 0, "N": 0, "false": 0, "False": 0, "0": 0
}
```

User Input Functions

```python
# Heart Input
def get_heart_input():

    print("\n❤️ HEART DISEASE PREDICTION")
    print("⚠ Enter values carefully (numbers only where required)\n")

    age = int(input("Age (e.g., 45): "))
    
    gender = gender_map[input("Gender (male/female): ")]
    cp = cp_map[input("Chest Pain (typical angina / atypical angina / non-anginal pain / asymptomatic): ")]

    bp = float(input("Resting Blood Pressure (e.g., 120): "))
    chol = float(input("Cholesterol (e.g., 200): "))

    fbs = yes_no[input("Fasting Blood Sugar >120? (yes/no): ")]
    max_hr = float(input("Max Heart Rate (e.g., 150): "))
    exercise = yes_no[input("Exercise induced angina? (yes/no): ")]

    return [
        age, gender, cp, bp, chol, fbs,
        0, max_hr, exercise, 1.0, 1, 0, 1
    ]
```

```python
# Diabetes Input
def get_diabetes_input():
    print("\n🩸 DIABETES RISK PREDICTION")
    print("⚠ Enter values carefully (numbers only where required)\n")
    
    pregnancies = int(input("Number of Pregnancies (Enter 0 if male or never pregnant): "))
    
    glucose = float(input("Glucose Level (2-hour plasma glucose concentration, e.g., 120): "))
    bp = float(input("Diastolic Blood Pressure (mm Hg, e.g., 80): "))
    skin_thickness = float(input("Triceps Skin Fold Thickness (mm, e.g., 20): "))
    insulin = float(input("2-Hour Serum Insulin (mu U/ml, e.g., 85): "))
    bmi = float(input("Body Mass Index (e.g., 32.5): "))
    dpf = float(input("Diabetes Pedigree Function (e.g., 0.47): "))
    age = int(input("Age (e.g., 33): "))

    return [
        pregnancies, glucose, bp, skin_thickness, 
        insulin, bmi, dpf, age
    ]
```

```python
# Breast Cancer Input
def get_cancer_input():

    print("\n🎗 BREAST CANCER PREDICTION")
    print("⚠ Enter values carefully (numbers only where required)\n")

    radius = float(input("Mean Radius (e.g., 14.0): "))
    texture = float(input("Mean Texture (e.g., 19.0): "))
    perimeter = float(input("Mean Perimeter (e.g., 92.0): "))
    area = float(input("Mean Area (e.g., 650.0): "))
    smoothness = float(input("Mean Smoothness (e.g., 0.1): "))

    # Fetch the dataset's baseline mean values for all 30 features
    default_features = cancer_df.drop("target", axis=1).mean().values

    # Overwrite the first 5 features with the user's explicit inputs
    default_features[0] = radius
    default_features[1] = texture
    default_features[2] = perimeter
    default_features[3] = area
    default_features[4] = smoothness

    # Convert from a NumPy array to a standard Python list to match other functions
    return list(default_features)
```

Unified Prediction Engine

```python
def predict(disease_type):

    if disease_type == "heart":
        data = get_heart_input()
        model = heart_model

    elif disease_type == "diabetes":
        data = get_diabetes_input()
        model = diabetes_model

    elif disease_type == "cancer":
        data = get_cancer_input()   
        model = cancer_model

    else:
        return "Invalid disease type"

    data = np.array(data).reshape(1, -1)
    return model.predict(data)[0]
```

Final System UI

```python
def run_system():

    print("\n🏥 MULTI-DISEASE AI SYSTEM")
    print("1. Heart Disease")
    print("2. Diabetes")
    print("3. Cancer")

    choice = input("Select option: ")

    if choice == "1":
        result = predict("heart")

        print("⚠ High Risk" if result == 1 else "✅ Low Risk")

    elif choice == "2":
        result = predict("diabetes")

        print("⚠ High Risk" if result == 1 else "✅ Low Risk")

    elif choice == "3":
        result = predict("cancer")

        print("⚠ High Risk" if result == 1 else "✅ Low Risk")

    else:
        print("Invalid choice")
```

```python
run_system()
```
