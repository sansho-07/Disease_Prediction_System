#Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
```

```python
# LOAD HEART DISEASE DATASET


heart_df = pd.read_csv(r"/Users/saansho/Shared/Files From d.localized/AI and ML projects/Disease_Prediction/data/raw/heart.csv")

print("Heart Dataset Loaded Successfully!")
print("Shape:", heart_df.shape)

heart_df.head()
```

```python
# LOAD DIABETES DATASET

diabetes_df = pd.read_csv(r"/Users/saansho/Shared/Files From d.localized/AI and ML projects/Disease_Prediction/data/raw/diabetes.csv")

print("Diabetes Dataset Loaded Successfully!")
print("Shape:", diabetes_df.shape)

diabetes_df.head()
```

```python
# LOAD BREAST CANCER DATASET

cancer = load_breast_cancer()

cancer_df = pd.DataFrame(
    cancer.data,
    columns=cancer.feature_names
)

cancer_df["target"] = cancer.target

print("Breast Cancer Dataset Loaded Successfully!")
print("Shape:", cancer_df.shape)

cancer_df.head()
```

```python
# CHECK DATASET INFORMATION

print("Heart Dataset Info")
print(heart_df.info())

print("\n============================\n")

print("Diabetes Dataset Info")
print(diabetes_df.info())

print("\n============================\n")

print("Cancer Dataset Info")
print(cancer_df.info())
```

```python
# CHECK MISSING VALUES

print("Heart Missing Values")
print(heart_df.isnull().sum())

print("\n============================\n")

print("Diabetes Missing Values")
print(diabetes_df.isnull().sum())

print("\n============================\n")

print("Cancer Missing Values")
print(cancer_df.isnull().sum())
```

```python
#Heart Disease

plt.figure(figsize=(6,4))

sns.countplot(x='target', data=heart_df)

plt.title("Heart Disease Distribution")

plt.show()
```

```python
#Diabetes

plt.figure(figsize=(6,4))

sns.countplot(x='Outcome', data=diabetes_df)

plt.title("Diabetes Distribution")

plt.show()
```

```python
#Breast Cancer

plt.figure(figsize=(6,4))

sns.countplot(x='target', data=cancer_df)

plt.title("Breast Cancer Distribution")

plt.show()
```
