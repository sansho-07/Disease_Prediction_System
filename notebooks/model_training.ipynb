```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import xgboost as xgb
import joblib
```

Load Processed Data

```python
# Heart
Xh_train = np.load("../data/processed/Xh_train.npy")
Xh_test = np.load("../data/processed/Xh_test.npy")
yh_train = np.load("../data/processed/yh_train.npy")
yh_test = np.load("../data/processed/yh_test.npy")

# Diabetes
Xd_train = np.load("../data/processed/Xd_train.npy")
Xd_test = np.load("../data/processed/Xd_test.npy")
yd_train = np.load("../data/processed/yd_train.npy")
yd_test = np.load("../data/processed/yd_test.npy")

# Cancer
Xc_train = np.load("../data/processed/Xc_train.npy")
Xc_test = np.load("../data/processed/Xc_test.npy")
yc_train = np.load("../data/processed/yc_train.npy")
yc_test = np.load("../data/processed/yc_test.npy")

print("Data loaded successfully!")
```

Model Training Function 

```python
def train_and_evaluate(model, X_train, X_test, y_train, y_test, name):
    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    acc = accuracy_score(y_test, pred)

    print(f"\n{name}")
    print("Accuracy:", acc)
    print(classification_report(y_test, pred))

    return model, acc
```

Heart Disease Models

```python
#Logistic Regression
lr = LogisticRegression()

lr_model, lr_acc = train_and_evaluate(
    lr, Xh_train, Xh_test, yh_train, yh_test, "Heart - Logistic Regression"
)
```

```python
#Random Forest
rf = RandomForestClassifier(n_estimators=200, random_state=42)

rf_model, rf_acc = train_and_evaluate(
    rf, Xh_train, Xh_test, yh_train, yh_test, "Heart - Random Forest"
)
```

```python
#SVM
svm = SVC()

svm_model, svm_acc = train_and_evaluate(
    svm, Xh_train, Xh_test, yh_train, yh_test, "Heart - SVM"
)
```

```python
#XGBoost
xgb_model = xgb.XGBClassifier(use_label_encoder=False, eval_metric="logloss")

xgb_model, xgb_acc = train_and_evaluate(
    xgb_model, Xh_train, Xh_test, yh_train, yh_test, "Heart - XGBoost"
)
```

Diabetes Models

```python
lr_d = LogisticRegression()
rf_d = RandomForestClassifier(n_estimators=200)
svm_d = SVC()
xgb_d = xgb.XGBClassifier(eval_metric="logloss")
```

```python
lr_d_model, lr_d_acc = train_and_evaluate(lr_d, Xd_train, Xd_test, yd_train, yd_test, "Diabetes - LR")
rf_d_model, rf_d_acc = train_and_evaluate(rf_d, Xd_train, Xd_test, yd_train, yd_test, "Diabetes - RF")
svm_d_model, svm_d_acc = train_and_evaluate(svm_d, Xd_train, Xd_test, yd_train, yd_test, "Diabetes - SVM")
xgb_d_model, xgb_d_acc = train_and_evaluate(xgb_d, Xd_train, Xd_test, yd_train, yd_test, "Diabetes - XGB")
```

Cancer Models


```python
lr_c = LogisticRegression()
rf_c = RandomForestClassifier(n_estimators=200)
svm_c = SVC()
xgb_c = xgb.XGBClassifier(eval_metric="logloss")
```

```python
lr_c_model, lr_c_acc = train_and_evaluate(lr_c, Xc_train, Xc_test, yc_train, yc_test, "Cancer - LR")
rf_c_model, rf_c_acc = train_and_evaluate(rf_c, Xc_train, Xc_test, yc_train, yc_test, "Cancer - RF")
svm_c_model, svm_c_acc = train_and_evaluate(svm_c, Xc_train, Xc_test, yc_train, yc_test, "Cancer - SVM")
xgb_c_model, xgb_c_acc = train_and_evaluate(xgb_c, Xc_train, Xc_test, yc_train, yc_test, "Cancer - XGB")
```

Model Comparison Table

```python
results = pd.DataFrame({
    "Model": ["LR", "RF", "SVM", "XGB"],
    "Heart Accuracy": [lr_acc, rf_acc, svm_acc, xgb_acc],
    "Diabetes Accuracy": [lr_d_acc, rf_d_acc, svm_d_acc, xgb_d_acc],
    "Cancer Accuracy": [lr_c_acc, rf_c_acc, svm_c_acc, xgb_c_acc],
})

results
```

Find Best Model

```python
best_heart_model = max(
    [(lr_model, lr_acc), (rf_model, rf_acc), (svm_model, svm_acc), (xgb_model, xgb_acc)],
    key=lambda x: x[1]
)[0]

print("Best Heart Model Selected!")
```
