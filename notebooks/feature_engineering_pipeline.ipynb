Importing Libraries

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
```

```python
heart_df = pd.read_csv("../data/raw/heart.csv")
diabetes_df = pd.read_csv("../data/raw/diabetes.csv")
cancer_df = pd.read_csv("../data/raw/breast_cancer.csv")
```

Feature Engineering

```python
heart_df["age_risk"] = heart_df["age"] > 50
heart_df["chol_risk"] = heart_df["chol"] > 240
```

```python
diabetes_df["glucose_risk"] = diabetes_df["Glucose"] > 140
diabetes_df["bmi_risk"] = diabetes_df["BMI"] > 30
```

```python
cancer_df["size_risk"] = cancer_df["mean radius"] > cancer_df["mean radius"].median()
```

Convert Boolean to int

```python
heart_df[["age_risk", "chol_risk"]] = heart_df[["age_risk", "chol_risk"]].astype(int)
diabetes_df[["glucose_risk", "bmi_risk"]] = diabetes_df[["glucose_risk", "bmi_risk"]].astype(int)
cancer_df["size_risk"] = cancer_df["size_risk"].astype(int)
```

Separate features and target

```python
X_heart = heart_df.drop("target", axis=1)
y_heart = heart_df["target"]

X_diabetes = diabetes_df.drop("Outcome", axis=1)
y_diabetes = diabetes_df["Outcome"]

X_cancer = cancer_df.drop("target", axis=1)
y_cancer = cancer_df["target"]
```

Building Pipeline

```python
#Heart Pipeline

heart_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
#Diabetes Pipeline

diabetes_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
##Cancer Pipeline

cancer_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

Train Test Split


```python
Xh_train, Xh_test, yh_train, yh_test = train_test_split(
    X_heart, y_heart, test_size=0.2, random_state=42
)

Xd_train, Xd_test, yd_train, yd_test = train_test_split(
    X_diabetes, y_diabetes, test_size=0.2, random_state=42
)

Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_cancer, y_cancer, test_size=0.2, random_state=42
)
```

Fit Pipelines

```python
Xh_train = heart_pipeline.fit_transform(Xh_train)
Xh_test = heart_pipeline.transform(Xh_test)
```

```python
Xd_train = diabetes_pipeline.fit_transform(Xd_train)
Xd_test = diabetes_pipeline.transform(Xd_test)
```

```python
Xc_train = cancer_pipeline.fit_transform(Xc_train)
Xc_test = cancer_pipeline.transform(Xc_test)
```
